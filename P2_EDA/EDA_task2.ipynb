```python
# ==========================================
# Exploratory Data Analysis (EDA)
# Dataset: Original Dataset for Data Analytics.xlsx
# ==========================================

# -----------------------------
# 1. Import Libraries
# -----------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# -----------------------------
# 2. Load Dataset
# -----------------------------
df = pd.read_excel("/content/Original Dataset for Data Analytics.xlsx")

print("Dataset Loaded Successfully!")
```

```python
# -----------------------------
# 3. Display Dataset
# -----------------------------
print("\nFirst 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())
```

```python
# -----------------------------
# 4. Dataset Shape
# -----------------------------
print("\nDataset Shape:")
print(df.shape)

print(f"\nRows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

```python
# -----------------------------
# 5. Column Names
# -----------------------------
print("\nColumns:")
print(df.columns)
```

```python
# -----------------------------
# 6. Dataset Information
# -----------------------------
print("\nDataset Information:")
print(df.info())
```

```python
# -----------------------------
# 7. Check Missing Values
# -----------------------------
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# -----------------------------
# 9. Descriptive Statistics
# -----------------------------
print("\nDescriptive Statistics:")
print(df.describe())
```

```python
# -----------------------------
# 10. Mean, Median & Mode
# -----------------------------
numeric_columns = df.select_dtypes(include=np.number).columns

print("\n========== Mean ==========")
print(df[numeric_columns].mean())

print("\n========== Median ==========")
print(df[numeric_columns].median())

print("\n========== Mode ==========")
print(df[numeric_columns].mode().iloc[0])
```

```python
# -----------------------------
# 11. Product Distribution
# -----------------------------
print("\nProduct Counts:")
print(df["Product"].value_counts())

plt.figure(figsize=(6,6))
df["Product"].value_counts().plot(kind="bar")
plt.title("Product Distribution")
plt.xlabel("Product")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# 12. Payment Method
# -----------------------------
print("\nPayment Method Counts:")
print(df["PaymentMethod"].value_counts())

plt.figure(figsize=(6,6))
df["PaymentMethod"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.title("Payment Methods")
plt.ylabel("")
plt.show()
```

```python
# -----------------------------
# 13. Order Status
# -----------------------------
print("\nOrder Status Counts:")
print(df["OrderStatus"].value_counts())

plt.figure(figsize=(6,6))
df["OrderStatus"].value_counts().plot(kind="bar")
plt.title("Order Status")
plt.xlabel("Status")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# 14. Referral Source
# -----------------------------
print("\nReferral Source Counts:")
print(df["ReferralSource"].value_counts())

plt.figure(figsize=(8,4))
df["ReferralSource"].value_counts().plot(kind="bar")
plt.title("Referral Source")
plt.xlabel("Source")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# 15. Histogram - Quantity
# -----------------------------
plt.figure(figsize=(6,4))
plt.hist(df["Quantity"], bins=10)
plt.title("Quantity Distribution")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()
print('-'*70)
# -----------------------------
# 16. Histogram - Unit Price
# -----------------------------
plt.figure(figsize=(6,4))
plt.hist(df["UnitPrice"], bins=20)
plt.title("Unit Price Distribution")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.show()
print('-'*70)
```

```python
# -----------------------------
# 17. Histogram - Total Price
# -----------------------------
plt.figure(figsize=(6,4))
plt.hist(df["TotalPrice"], bins=20)
plt.title("Total Price Distribution")
plt.xlabel("Total Price")
plt.ylabel("Frequency")
plt.show()
```

```python
# -----------------------------
# 18. Boxplots (Outlier Detection)
# -----------------------------
plt.figure(figsize=(6,4))
plt.boxplot(df["Quantity"])
plt.title("Outliers in Quantity")
plt.show()
print('-'*70)
plt.figure(figsize=(6,4))
plt.boxplot(df["UnitPrice"])
plt.title("Outliers in Unit Price")
plt.show()
print('-'*70)
plt.figure(figsize=(6,4))
plt.boxplot(df["TotalPrice"])
plt.title("Outliers in Total Price")
plt.show()
```

```python
# -----------------------------
# 19. Correlation Matrix
# -----------------------------
correlation = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(8,6))
plt.imshow(correlation, cmap="coolwarm")

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.colorbar()

for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(j, i,
                 round(correlation.iloc[i, j], 2),
                 ha='center',
                 va='center',
                 color='black',
                 fontsize=8)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# 20. Top 10 Highest Orders
# -----------------------------
print("\nTop 10 Highest Total Price Orders:")
print(df.sort_values("TotalPrice", ascending=False).head(10))
```

```python
# -----------------------------
# 21. Coupon Usage
# -----------------------------
print("\nCoupon Code Usage:")
print(df["CouponCode"].value_counts(dropna=False))
```

```python
# -----------------------------
# 22. Orders by Product
# -----------------------------
orders_by_product = df.groupby("Product")["TotalPrice"].sum()

print("\nRevenue by Product:")
print(orders_by_product)

plt.figure(figsize=(8,5))
orders_by_product.plot(kind="bar")
plt.title("Revenue by Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# 23. Average Total Price by Payment Method
# -----------------------------
avg_payment = df.groupby("PaymentMethod")["TotalPrice"].mean()

print("Average Total Price by Payment Method:")
print(avg_payment)

plt.figure(figsize=(7,4))
avg_payment.plot(kind="bar")
plt.title("Average Total Price by Payment Method")
plt.ylabel("Average Price")
plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# 24. Average Total Price by Order Status
# -----------------------------
avg_status = df.groupby("OrderStatus")["TotalPrice"].mean()

print("\nAverage Total Price by Order Status:")
print(avg_status)

plt.figure(figsize=(6,4))
avg_status.plot(kind="bar")
plt.title("Average Total Price by Order Status")
plt.ylabel("Average Price")
plt.tight_layout()
plt.show()
```

```python
# -----------------------------
# 25. Final Summary
# -----------------------------
print("==============================")
print("EDA SUMMARY")
print("==============================")
print(f"Total Rows : {df.shape[0]}")
print(f"Total Columns : {df.shape[1]}")
print(f"Missing Values :\n{df.isnull().sum()}")
print(f"\nDuplicate Rows : {df.duplicated().sum()}")

print("\nMean Values:")
print(df[numeric_columns].mean())

print("\nMedian Values:")
print(df[numeric_columns].median())

print("\nEDA Completed Successfully!")
```
